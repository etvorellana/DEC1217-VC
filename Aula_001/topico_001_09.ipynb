# COMPARANDO CATEGORIAS

Os gráficos deste capítulo visam ajudar nossos leitores a comparar valores entre categorias. Barras, linhas e pontos permitem que os leitores façam comparações tanto dentro de grupos quanto entre eles. Em alguns casos, queremos que o leitor visualize tanto os níveis quanto as variações, ou alguma outra combinação de variáveis; em outros, queremos concentrar sua atenção em uma comparação específica.

O desafio ao comparar dados categóricos é decidir o que queremos que o gráfico transmita. Existe um argumento ou uma narrativa principal? Há algum aspecto que possa ser identificado como a comparação mais importante que se deseja que o leitor faça? Como criadores de gráficos, precisamos priorizar a função que queremos que eles desempenhem. Ao incluir todas as barras ou pontos no gráfico, podemos acabar obscurecendo a mensagem que desejamos transmitir.

Este capítulo começa com o gráfico de barras. Assim como o gráfico de linhas que dará início ao próximo capítulo, o gráfico de barras é familiar à maioria dos leitores, o que o torna uma escolha conveniente para orientá-los ao comparar categorias ou observar mudanças ao longo do tempo. Ele também ocupa o topo do diagrama de hierarquia de percepção visual. Não é obrigatório fornecer sempre os valores exatos aos leitores, mas, quando o fazemos, o gráfico de barras é uma excelente escolha.



## GRÁFICOS GAUGE (DE MEDIDOR OU VELOCÍMETRO) E BULLET (DE BALA OU MARCADORES)

O gráfico de medidor (ou diagrama de medidor) assemelha-se ao velocímetro do painel de um carro. Geralmente configurado em um formato que varia de meio círculo a um círculo completo, ele utiliza um ponteiro ou agulha para indicar a posição dos dados dentro de uma determinada faixa. As seções do medidor são sombreadas para ilustrar categorias como ruim, bom e excelente.

Vejo gráficos de medidor (tipo velocímetro) com mais frequência em ferramentas de planejamento financeiro, pois oferecem uma maneira simples e familiar de visualizar metas ou o progresso em direção a um objetivo. Eles também aparecem frequentemente em campanhas de arrecadação de fundos, nas quais o semicírculo completo representa a meta, enquanto o ponteiro e a área preenchida indicam o valor arrecadado até o momento. Esse é um bom exemplo do uso de uma forma familiar para reforçar a metáfora da visualização: todos entendem que "preencher" o medidor significa que a campanha atingiu sua meta.

No entanto, os gráficos de medidor (*gauge charts*) apresentam desafios de percepção, pois, mais uma vez, as pessoas não são muito boas em medir e comparar ângulos. Se você deseja transmitir ao leitor uma noção geral dos valores, o gráfico de medidor é uma boa escolha. Mas, se for fundamental permitir que o leitor identifique valores específicos e os compare com as faixas de referência, então ele não é a opção ideal.

Dada a familiaridade com o indicador de ponteiro (tipo manômetro) e a metáfora óbvia que ele representa, talvez não surpreenda que eles apareçam em diversos contextos. Como exemplo, certa vez recebi um folheto da Mountjoy Properties — uma equipe do setor imobiliário residencial que atua na região metropolitana de Washington, D.C. — contendo uma série de gráficos desse tipo que mostravam as tendências imobiliárias atuais no meu bairro, no norte da Virgínia. Consegui visualizar rapidamente a situação do mercado, mas a inclusão de mais dados ou de informações mais detalhadas talvez fosse mais difícil.




### GRÁFICOS DE BARRA TIPO BULLET

Devido a esses desafios de percepção associados ao gráfico de medidor (*gauge chart*), o autor Stephen Few criou o gráfico de bala (*bullet chart*), que é uma forma linear e mais compacta de exibir tipos semelhantes de dados. O gráfico de bala básico contém três elementos de dados distintos:

1. Primeiramente, há o valor real ou observado, representado aqui pela barra horizontal preta. Nesta ilustração, a barra indica uma pontuação média de satisfação do cliente de 4,0.

2. Em segundo lugar, há o valor-alvo, representado aqui pela linha vertical preta. Neste caso, a meta era uma pontuação de satisfação de 3,5.

3. Por fim, há a faixa de fundo, que mostra níveis ou categorias de desempenho, como insatisfatório, bom e excelente. Esses elementos ficam posicionados atrás das outras duas séries, permitindo ao leitor comparar os valores reais com os valores-alvo. Aqui, pontuações de 1,5 ou menos são consideradas insatisfatórias; de 1,5 a 3,0, boas; e acima de 3,0, excelentes.

Os diferentes componentes dos gráficos de barra de progresso (*bullet charts*) podem variar. Pode haver uma escala com cinco faixas em vez de três, ou talvez não exista um valor de referência (meta). As escalas também podem exibir a distribuição subjacente dos dados — por exemplo, mostrando quartis ou faixas de percentis (veja o quadro na página 183 para mais informações sobre percentis). Por serem muito compactos, é fácil criar várias versões desses gráficos e empilhá-los. Os gráficos acima apresentam três métricas comuns em relatórios financeiros; no entanto, eles são mais compactos do que os gráficos de medidor (*gauge charts*) e, como os retângulos estão alinhados, facilitam a comparação entre as diferentes categorias.
