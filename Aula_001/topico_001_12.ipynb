# COMPARANDO CATEGORIAS

Os gráficos deste capítulo visam ajudar nossos leitores a comparar valores entre categorias. Barras, linhas e pontos permitem que os leitores façam comparações tanto dentro de grupos quanto entre eles. Em alguns casos, queremos que o leitor visualize tanto os níveis quanto as variações, ou alguma outra combinação de variáveis; em outros, queremos concentrar sua atenção em uma comparação específica.

O desafio ao comparar dados categóricos é decidir o que queremos que o gráfico transmita. Existe um argumento ou uma narrativa principal? Há algum aspecto que possa ser identificado como a comparação mais importante que se deseja que o leitor faça? Como criadores de gráficos, precisamos priorizar a função que queremos que eles desempenhem. Ao incluir todas as barras ou pontos no gráfico, podemos acabar obscurecendo a mensagem que desejamos transmitir.

Este capítulo começa com o gráfico de barras. Assim como o gráfico de linhas que dará início ao próximo capítulo, o gráfico de barras é familiar à maioria dos leitores, o que o torna uma escolha conveniente para orientá-los ao comparar categorias ou observar mudanças ao longo do tempo. Ele também ocupa o topo do diagrama de hierarquia de percepção visual. Não é obrigatório fornecer sempre os valores exatos aos leitores, mas, quando o fazemos, o gráfico de barras é uma excelente escolha.



## GRÁFICO DE CASCATA

Um gráfico de cascata ilustra uma equação matemática básica: a adição ou subtração de valores a partir de um valor inicial para chegar a um montante final. Trata-se, essencialmente, de um gráfico de barras, mas cada barra subsequente começa onde a anterior terminou, mostrando como os valores se acumulam ao longo do gráfico. Normalmente, valores negativos recebem uma cor diferente dos positivos, assim como os totais de início e fim. A inclusão de linhas conectando as barras pode orientar o leitor na visualização. Como essas linhas servem de guia e não representam dados propriamente ditos, devem ser mais claras e finas do que os demais elementos.

O próximo gráfico mostra as contribuições para a renda bruta total e a renda líquida total da Austrália em 2016. Os dados são os mesmos utilizados no exemplo de mapa de calor anterior; no entanto, naquela abordagem, foi possível exibir dados de dez países em uma única visualização. Imagine tentar fazer o mesmo com um gráfico de cascata: precisaríamos de dez gráficos diferentes — algo que pode ser útil em certos cenários, mas que certamente é menos compacto do que o mapa de calor.

Gráficos de cascata também podem exibir variações ao longo do tempo. É possível mostrar, por exemplo, as contribuições para o PIB total de um ano para o outro e como diferentes valores contribuem para a variação ao longo do ano. Qualquer série de dados que seja somada ou subtraída de outra pode ser apresentada dessa forma; no entanto, vale ressaltar que se trata de um tipo de gráfico não convencional, o que pode exigir mais tempo do leitor para compreendê-lo.

## CONCLUSÃO

De barras individuais a grupos de barras e barras empilhadas, o gráfico de barras é uma das formas mais familiares de visualização de dados para exibir comparações categóricas. Ele também ocupa o topo da nossa escala de hierarquia perceptiva mencionada anteriormente. No entanto, o gráfico de barras apresenta certos desafios: um número excessivo de barras pode tornar a visualização visualmente sobrecarregada e confusa, e o empilhamento de séries dificulta a comparação daquelas que não estão alinhadas ao longo do mesmo eixo.

O formato básico de barra pode ser organizado de diversas maneiras. As barras podem ficar lado a lado ou divergir de uma linha de base central. Podem ser empilhadas umas sobre as outras, seja na horizontal ou na vertical — ou em ambas as direções, como em um gráfico de mosaico. Também podem ser dispostas para representar equações matemáticas simples, como ocorre em um gráfico de cascata. De modo geral, temos facilidade em identificar valores de dados com base no comprimento das barras; por isso, muitos desses tipos de gráfico permitem que o leitor perceba facilmente o valor exato.

Existem outras maneiras de permitir que o leitor faça comparações. Gosto particularmente dos gráficos de pontos (*dot plots*), pois eles eliminam o excesso de tinta característico dos gráficos de barras convencionais e liberam espaço para a inclusão de anotações e rótulos. O uso de ícones, quadrados ou outras formas pode envolver o público de maneiras que os gráficos tradicionais talvez não consigam, embora possam resultar em uma menor densidade de dados.

Embora os gráficos de barras ocupem o topo da hierarquia de percepção visual, sejamos francos: eles podem ser muito monótonos. Vemos gráficos de barras todos os dias. Como criadores de gráficos, nosso desafio é, por vezes, encontrar formas de envolver o público — e recorrer a tipos de gráficos menos usuais do nosso repertório de visualização de dados pode justamente alcançar esse objetivo. Cabe a nós decidir onde queremos concentrar a atenção do leitor: no nível ou na diferença, na comparação individual ou múltipla, ou nos valores relativos ou totais.
