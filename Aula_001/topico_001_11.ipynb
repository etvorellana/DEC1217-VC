# COMPARANDO CATEGORIAS

Os gráficos deste capítulo visam ajudar nossos leitores a comparar valores entre categorias. Barras, linhas e pontos permitem que os leitores façam comparações tanto dentro de grupos quanto entre eles. Em alguns casos, queremos que o leitor visualize tanto os níveis quanto as variações, ou alguma outra combinação de variáveis; em outros, queremos concentrar sua atenção em uma comparação específica.

O desafio ao comparar dados categóricos é decidir o que queremos que o gráfico transmita. Existe um argumento ou uma narrativa principal? Há algum aspecto que possa ser identificado como a comparação mais importante que se deseja que o leitor faça? Como criadores de gráficos, precisamos priorizar a função que queremos que eles desempenhem. Ao incluir todas as barras ou pontos no gráfico, podemos acabar obscurecendo a mensagem que desejamos transmitir.

Este capítulo começa com o gráfico de barras. Assim como o gráfico de linhas que dará início ao próximo capítulo, o gráfico de barras é familiar à maioria dos leitores, o que o torna uma escolha conveniente para orientá-los ao comparar categorias ou observar mudanças ao longo do tempo. Ele também ocupa o topo do diagrama de hierarquia de percepção visual. Não é obrigatório fornecer sempre os valores exatos aos leitores, mas, quando o fazemos, o gráfico de barras é uma excelente escolha.



## DIAGRAMA SANKEY

Os diagramas de Sankey — nomeados em homenagem ao seu criador, Matthew Henry Phineas Riall Sankey, em 1898 — são particularmente úteis para mostrar como categorias se comparam entre si e como fluem para outros estados ou categorias. Setas ou linhas exibem a transição de um estado para outro, e a largura das linhas indica a magnitude de cada transição. As mudanças podem ocorrer ao longo do tempo ou na forma de comparações entre categorias. Por exemplo, um diagrama de Sankey poderia ser utilizado para mostrar como diferentes empresas de um setor se fundiram, se desmembraram ou faliram ao longo de diferentes anos.

Este diagrama de Sankey mostra como cinquenta e dois alunos tentaram soletrar a palavra "camouflage". O primeiro segmento azul mostra que todos os cinquenta e dois alunos começaram com a letra "C"; em seguida, cinquenta escreveram "Cam", seguidos por trinta e sete que escreveram "Camof", e assim por diante. Dez alunos escreveram a palavra corretamente, o que é indicado pelo segmento laranja próximo ao topo do gráfico.

Mais pertinente ao tipo de conteúdo que analisamos até agora neste capítulo, o gráfico acima mostra os fluxos de ajuda federal dos Estados Unidos, do Reino Unido e da Alemanha para regiões de todo o mundo. É possível observar que os países da África Subsaariana e países não especificados recebem a maior parte da ajuda, e que o Reino Unido e os Estados Unidos contribuem mais para os países da África Subsaariana do que a Alemanha.

Apresentar esses dados em gráficos de barras pareadas ou empilhadas nos oferece uma perspectiva diferente. No gráfico de barras pareadas à esquerda, a primeira comparação óbvia é entre os países financiadores para cada região do mundo. Já no gráfico de barras empilhadas, a tendência é comparar o financiamento entre as regiões receptoras — observando, por exemplo, que a maior parte dos recursos destinados aos países da África Subsaariana provém dos Estados Unidos. O diagrama de Sankey chama a atenção de outra forma: sua leitura ocorre na horizontal, combinando essas duas comparações sem privilegiar nenhuma delas. Nenhuma das visualizações está "certa" ou "errada"; elas podem atender a públicos distintos, destacar padrões diferentes e responder a perguntas variadas.

Diagramas de Sankey podem ser sobrepostos a outros tipos de gráfico. Neste exemplo, os fluxos de ajuda provenientes dos Estados Unidos são apresentados em um mapa-múndi. Isso oferece uma perspectiva geográfica dos dados, mas também simplifica a visualização — imagine como o gráfico ficaria poluído se incluísse fluxos dos Estados Unidos, do Reino Unido e da Alemanha.

O maior problema dos diagramas de Sankey — e de muitos outros gráficos, aliás — é a representação de um número excessivo de séries, como ocorre nesta versão que inclui mais países. Com muitos grupos ou cruzamentos, o gráfico torna-se difícil de interpretar. Se você se deparar com um excesso de linhas ou cruzamentos, tente simplificar os dados, utilizar múltiplos diagramas de Sankey ou optar por outro tipo de gráfico.
