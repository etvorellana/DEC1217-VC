# COMPARANDO CATEGORIAS

Os gráficos deste capítulo visam ajudar nossos leitores a comparar valores entre categorias. Barras, linhas e pontos permitem que os leitores façam comparações tanto dentro de grupos quanto entre eles. Em alguns casos, queremos que o leitor visualize tanto os níveis quanto as variações, ou alguma outra combinação de variáveis; em outros, queremos concentrar sua atenção em uma comparação específica.

O desafio ao comparar dados categóricos é decidir o que queremos que o gráfico transmita. Existe um argumento ou uma narrativa principal? Há algum aspecto que possa ser identificado como a comparação mais importante que se deseja que o leitor faça? Como criadores de gráficos, precisamos priorizar a função que queremos que eles desempenhem. Ao incluir todas as barras ou pontos no gráfico, podemos acabar obscurecendo a mensagem que desejamos transmitir.

Este capítulo começa com o gráfico de barras. Assim como o gráfico de linhas que dará início ao próximo capítulo, o gráfico de barras é familiar à maioria dos leitores, o que o torna uma escolha conveniente para orientá-los ao comparar categorias ou observar mudanças ao longo do tempo. Ele também ocupa o topo do diagrama de hierarquia de percepção visual. Não é obrigatório fornecer sempre os valores exatos aos leitores, mas, quando o fazemos, o gráfico de barras é uma excelente escolha.



## CCOMPARAÇÃO DE BOLHAS E BOLHAS ANINHADAS

No gráfico de bolhas básico, círculos representam valores. Assim como no gráfico de barras, o objetivo desses gráficos é comparar valores entre categorias. No entanto, diferentemente do gráfico de barras, os seres humanos não são muito bons em comparar com precisão o tamanho de círculos (lembre-se do diagrama de hierarquia de percepção da página 14). Ainda assim, os círculos podem ser visualmente mais interessantes, reforçar um elemento visual ou uma metáfora e constituir uma boa escolha quando identificar quantidades exatas não for o aspecto mais importante.

Outra desvantagem dos círculos de tamanho proporcional é a impossibilidade de visualizar valores negativos. Enquanto as barras podem se estender em ambas as direções — tipicamente para a direita ou para cima, no caso de valores positivos, e para a esquerda ou para baixo, no caso de negativos —, essa visualização é muito mais difícil com círculos.

De qualquer forma, não somos muito bons em fazer estimativas precisas a partir de círculos, mesmo quando o tamanho deles reflete a área. Em vez disso, acredito que tentamos fazer a comparação com base no diâmetro do círculo — como em um gráfico de barras —, o que nos leva a conclusões incorretas. Observe os dois gráficos na página seguinte e tente estimar a porcentagem de pessoas que vivem com menos de US$ 1,90 por dia no Paquistão. Você acha que essa tarefa é mais fácil no gráfico de bolhas ou no gráfico de barras?

Isso não significa que você nunca deva usar círculos. Lembre-se, mais uma vez, de que tudo gira em torno do seu público. Um gráfico de comparação de bolhas, inserido em um artigo curto e posicionado lateralmente — com os números em destaque no centro de cada círculo —, pode ser mais atraente do que um gráfico de barras convencional. No entanto, o excesso de círculos pode dificultar a percepção de quantidades ou relações pelo seu público. No exemplo a seguir, é possível ver que a Índia, a República Democrática do Congo e a Nigéria apresentam os maiores contingentes populacionais em situação de pobreza; contudo, é difícil avaliar rapidamente a magnitude dessas diferenças ou os números referentes ao grupo seguinte de países.

Se utilizássemos uma representação visualmente mais precisa desses mesmos dados em, digamos, um gráfico de barras, o elemento visual ficaria muito maior, como na página anterior. Aqui, cada barra está identificada, permitindo que o leitor localize Madagascar; mas será que isso é importante? Mais uma vez: o objetivo é exibir todos os países ou apenas um subconjunto? Como sempre, leve em conta seus objetivos e se o leitor realmente precisa de uma representação visualmente precisa para compreender seu argumento.

Os gráficos de bolhas mostrados acima são conhecidos como gráficos de comparação de bolhas. A sobreposição de círculos, como no gráfico acima, é frequentemente chamada de gráfico de bolhas aninhadas. O gráfico de bolhas aninhadas pode, às vezes, ocultar círculos que estão ao fundo, mas também pode facilitar as comparações.

Você pode usar bolhas para demonstrar correlações (veja o gráfico de bolhas no Capítulo 8) ou adicioná-las a um mapa para representar outra variável (veja o mapa de pontos no Capítulo 7). De modo geral, embora o uso de círculos e bolhas na visualização de dados apresente questões relacionadas à percepção, eles também podem ser mais envolventes e agradáveis ​​do que mais um gráfico de barras ou de linhas. Como disse Amanda Cox, editora de dados do *New York Times*: "Há uma vertente no mundo da visualização de dados que defende que tudo poderia ser um gráfico de barras. Isso pode até ser verdade, mas também pode resultar em um mundo sem graça".
