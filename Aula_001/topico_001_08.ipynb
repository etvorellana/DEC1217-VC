# COMPARANDO CATEGORIAS

Os gráficos deste capítulo visam ajudar nossos leitores a comparar valores entre categorias. Barras, linhas e pontos permitem que os leitores façam comparações tanto dentro de grupos quanto entre eles. Em alguns casos, queremos que o leitor visualize tanto os níveis quanto as variações, ou alguma outra combinação de variáveis; em outros, queremos concentrar sua atenção em uma comparação específica.

O desafio ao comparar dados categóricos é decidir o que queremos que o gráfico transmita. Existe um argumento ou uma narrativa principal? Há algum aspecto que possa ser identificado como a comparação mais importante que se deseja que o leitor faça? Como criadores de gráficos, precisamos priorizar a função que queremos que eles desempenhem. Ao incluir todas as barras ou pontos no gráfico, podemos acabar obscurecendo a mensagem que desejamos transmitir.

Este capítulo começa com o gráfico de barras. Assim como o gráfico de linhas que dará início ao próximo capítulo, o gráfico de barras é familiar à maioria dos leitores, o que o torna uma escolha conveniente para orientá-los ao comparar categorias ou observar mudanças ao longo do tempo. Ele também ocupa o topo do diagrama de hierarquia de percepção visual. Não é obrigatório fornecer sempre os valores exatos aos leitores, mas, quando o fazemos, o gráfico de barras é uma excelente escolha.



## MAPA DE CALOR

Mapas de calor utilizam cores e níveis de saturação de cor para representar valores de dados. De forma simplificada, um mapa de calor é uma tabela com células codificadas por cores. Eles são frequentemente utilizados para visualizar dados de alta frequência ou quando identificar padrões gerais é mais importante do que conhecer valores exatos.

Estes dois mapas de calor mostram os diferentes componentes da renda total de dez países, utilizando dados do *Luxembourg Income Study*. A versão à esquerda utiliza a mesma escala de cores para todas as seis categorias, na qual cores mais claras representam valores menores e cores mais escuras representam valores maiores. Nessa visualização, percebe-se que os rendimentos do trabalho constituem a maior parcela da renda total e que, na maioria dos países, os benefícios sociais públicos representam a segunda maior parcela. No mapa de calor à direita, cada categoria possui sua própria escala de cores. Aqui, é possível observar mais claramente que os benefícios sociais públicos (na quarta coluna) representam uma parcela menor da renda total na Austrália, no Brasil, na Suíça e nos Estados Unidos. Qual deles é melhor depende dos seus objetivos. Você quer que o leitor compare todos os valores entre si ou faça comparações dentro de cada categoria?

Você também pode usar um mapa de calor para mostrar mudanças ao longo do tempo. Imagine uma planilha que contenha as taxas de infecção por sarampo em todos os estados dos EUA, de 1928 a 2008. Se a planilha apresentasse os estados organizados nas linhas e os anos nas colunas, seu primeiro impulso poderia ser criar o gráfico de linhas mostrado na página seguinte.

Não há nada inerentemente errado com o gráfico de linhas a seguir: é possível observar a taxa de infecções de 1928 até cerca de 1963 (marcado pela linha vertical preta), ano em que a vacina contra o sarampo foi introduzida. Nos cerca de cinco anos seguintes, as infecções caíram rapidamente e, por fim, chegaram a um patamar próximo de zero em um período de aproximadamente dez anos. Em suma, o que se depreende desse gráfico é que houve infecções — com oscilações de alta e baixa, em um emaranhado de linhas — até por volta de 1963.

O Wall Street Journal analisou a mesma planilha e, em vez de criar um gráfico de linhas denso, elaborou um mapa de calor. Criei minha própria versão aqui, utilizando uma paleta de cores diferente e categorias discretas para a taxa de infecção. É possível observar as células em azul mais escuro (predominantemente acima de 16 infecções por 100.000 habitantes) antes da introdução da vacina, também marcadas por uma linha preta. Após 1963, as cores mudam rapidamente para tons de azul mais claros e, por fim, para as taxas de infecção mais baixas (zero infecções e menos de 1 infecção por 100.000 habitantes).

Este gráfico pode não ser inerentemente melhor do que o gráfico de linhas, mas permite examinar cada estado ou ano com muito mais facilidade do que tentar identificar uma única linha em meio à confusão visual de um gráfico de linhas. Lembre-se também de que, às vezes, ser diferente é algo positivo por si só. Quantos gráficos de linhas complexos você já viu e simplesmente ignorou de imediato? O mapa de calor, com seu visual e cores distintos, pode atrair a atenção dos leitores. Como disse a artista e especialista em visualização de dados Giorgia Lupi: "A beleza é um ponto de entrada fundamental para que os leitores se interessem pela visualização e queiram explorá-la mais a fundo. A beleza não substitui a funcionalidade, mas, quando combinadas, ambas alcançam resultados melhores".

Outra forma de utilizar o mapa de calor é modificar o layout, por exemplo, aplicando-o a um calendário. Neste exemplo, as mortes em acidentes envolvendo veículos em 2015 são representadas em mapas de calor organizados por mês. Observe como é fácil identificar a maior taxa de mortalidade às sextas e sábados, ao longo da borda direita de cada coluna.

Para fins de comparação, considere plotar os mesmos dados em um gráfico de linhas. Mesmo com os círculos azuis adicionais usados ​​para marcar os sábados, é difícil chegar à mesma conclusão sobre a maior ocorrência de fatalidades nos fins de semana.

Ao contrário do exemplo do sarampo, em que ambos os gráficos apresentavam vantagens e desvantagens, neste caso o mapa de calor em formato de calendário é uma abordagem melhor, pois destaca com mais eficácia o padrão importante de mortes nos fins de semana — além de ser um gráfico mais envolvente, apresentado em um formato familiar.

Uma última maneira de modificar o mapa de calor é alterar completamente o layout retangular. Na página seguinte, cada um dos cinquenta estados é plotado ao longo dos raios de um círculo, agrupados em cinco regiões geográficas (separadas pelas linhas pretas). Cada anel representa um tipo de dado diferente (binário), como, por exemplo, se o estado possui leis de "direito ao trabalho" (*right-to-work*), imposto de renda ou salário mínimo.

Essa alternativa a um conjunto de seis mapas apresenta vantagens e desvantagens. Por um lado, trata-se de uma representação compacta de seis séries de dados diferentes, permitindo ao leitor visualizar de forma rápida e clara dados categóricos e de relação parte-todo, tanto dentro de cada estado quanto entre eles. Por outro lado, não é um tipo de gráfico familiar, o que pode acabar afastando alguns leitores. Vale ressaltar também que a ordem dos anéis pode influenciar a nossa percepção dos dados, uma vez que os quadrados (vermelhos) do anel externo são, por definição, maiores do que os quadrados (amarelos) do anel interno.
